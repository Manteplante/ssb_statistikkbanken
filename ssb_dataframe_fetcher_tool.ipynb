Inspect table in detail

```python
def inspect_table(table_id, time_period='top(1)', sample=True):
    
    print(f"\n{'='*70}")
    print(f"TABLE INSPECTION: {table_id}")
    print(f"{'='*70}\n")
    
    fetcher.print_table_info(table_id)
    
    print("AVAILABLE FILTERS:")
    print("-" * 70)
    fetcher.print_dimensions(table_id)
    
    if sample:
        print("SAMPLE DATA (with filters applied):")
        print("-" * 70)
        df_sample = fetcher.fetch_clean_table(table_id, time_period=time_period)
        print(f"Shape: {df_sample.shape}")
        print(f"Columns: {list(df_sample.columns)}\n")
        print(df_sample.head(10))
    
    print(f"\nHOW TO USE EXAMPLE:")
    print("-" * 70)
    print("df = fetcher.fetch_clean_table(")
    print(f"    '{table_id}',")
    print("    time_period='top(10)',")
    print("    filters={'DimensionName': 'value_code'}")
    print(")")
    print(f"{'='*70}\n")

inspect_table('05810')
```
